## 1. What are we building?

We will create a simple AI Tutor Agent.

The Agent will use Gemini as its LLM and will be given a task to explain an AI concept.

### Mental model

Think of the components this way:

| Component | Responsibility |
|---|---|
| LLM | Provides the language/reasoning capability |
| Agent | Defines who performs the work |
| Task | Defines what needs to be done |
| Crew | Orchestrates the work |

The execution flow is:

```text
Crew
  ↓
Task
  ↓
Agent
  ↓
Gemini LLM
  ↓
Gemini API
  ↓
Response

## 3. Configure Gemini

CrewAI provides the orchestration layer, but the actual language generation is performed by an LLM.

For this lab, Gemini is the LLM.

```text
CrewAI
   ↓
Agent
   ↓
Gemini

````python
from crewai import LLM

llm = LLM(
    model=# Use a Gemini model available to your Google AI Studio project.,
    api_key=os.getenv("GEMINI_API_KEY")
)
print(llm)
````

## 4. Create an Agent

An Agent represents a worker with a specific role, goal and behavior.

A useful mental model is:

```text
Agent
├── Role
├── Goal
├── Backstory / behavior
├── LLM
└── Tools (introduced later)

````python
from crewai import Agent

gemini_agent = Agent(
    role="AI Tutor",
    goal="Explain AI concepts clearly to beginners",
    backstory="You are a patient AI teacher who explains concepts step by step.",
    llm=llm,
    verbose=True
)
print("Agent created:", gemini_agent.role)
````

## 5. Create a Task

The Agent describes **who** is doing the work.

The Task describes **what** needs to be done.

```text
Agent → WHO?
Task  → WHAT?

````python
from crewai import Task

gemini_task = Task(
    description="Explain what an AI agent is in simple terms.",
    expected_output="A beginner-friendly explanation with a simple example.",
    agent=gemini_agent
)
````

## 6. Create a Crew

A Crew brings Agents and Tasks together.

For this first example we only have:

- 1 Agent
- 1 Task

Later, a Crew can coordinate multiple Agents performing different Tasks.

````python
from crewai import Crew

gemini_crew = Crew(
    agents=[gemini_agent],
    tasks=[gemini_task]
)
````

## 7. Execute the Crew

Jupyter already runs an asynchronous event loop.

Therefore, when executing CrewAI from a notebook, use the asynchronous API:

```python
await crew.kickoff_async()

````python
result = await gemini_crew.kickoff_async()

print(result)
````
